```python
## OBS Koden är INTE user proof. Om du skriver dumheter får du dumheter som return.
  ## RE: Vissa är säkrade nu.



import random
from random import randrange
import math
from IPython.display import clear_output
from time import sleep


# Kraschsäkring där ett heltal används.
def input_int(prompt):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            delay_print("...maybe try writing a number?")

# Snyggare print kommando, skriver ut en bokstav i taget, gör spelet lite mer intressant och UI-design snyggare (jag tycker att estetiken passar bättre).
def delay_print(text):
    for letter in text:
        print(letter, end="", flush=True)
        sleep(0.04)
    print()


# Spelsekvensen
def gameplay_loop():
  global blank
  global chamber
  delay_print(f"???:       In the gun there are {bullet} bullet(s) and {blank} blank(s) arranged in an unknown order.")
  delay_print(f"???:       What do you wish to do?\n\n [1] Shoot yourself\n [2] Shoot opponent\n")
  choice = input_int("> ")

  ## chance = bullet / total -användaren förväntas tänka på detta medan den spelar för att ta reda på vad hen bör välja.

  ## Logistik

  if choice == 1:
    current_round = chamber.pop(0)
    delay_print(f"Very brave")
    delay_print(f"You point the gun at yourself...")
    sleep(1.4)
    if current_round == "live":
      delay_print(f"...BANG")
      delay_print(f"\033[91m --- you died... ---\033[0m")
      sleep(1.4)
      clear_output()
      delay_print(f"Do you want to try again? y/n")
      retry = input("> ")
      if retry == "y":
        gameplay_loop()
      elif retry == "n":
        exit()
    else:
      delay_print("...CLICK! It was a blank")
      blank = blank - 1
      total = blank + bullet
      chamber = ["live"] * bullet + ["blank"] * blank
      delay_print(f"You get another chance. There are {total} bullets left in the chamber.")
      gameplay_loop()

  elif choice == 2:
    current_round = chamber.pop(0)
    delay_print(f"Cowardly... or clever?")
    delay_print(f"You point the gun at the opponent...")
    sleep(1.4)
    if current_round == "live":
      delay_print(f"...BANG! You shot the opponent.")
      delay_print(f"CONGRATULATIONS! You win!")
    else:
      delay_print(f"...CLICK! It was a blank. Opponent's turn.\n")
      sleep(1.4)
      blank = blank - 1
      total = blank + bullet
      chamber = ["live"] * bullet + ["blank"] * blank
      random.shuffle(chamber)
      current_round = chamber.pop(0)
      if current_round == "live":
        delay_print(f"...BANG!")
        delay_print(f"\033[91m --- you died... ---\033[0m")
        sleep(1.4)
        clear_output()
        delay_print(f"Would you like to try again? y/n")
        retry = input("> ")
        if retry == "y":
          clear_output()
          gameplay_loop()
        elif retry == "n":
          exit()
      elif current_round == "blank":
        delay_print(f"...CLICK! It was a blank. {name} your turn.\n")
        blank = blank - 1
        total = blank + bullet
        chamber = ["live"] * bullet + ["blank"] * blank
        random.shuffle(chamber)
        sleep(1.4)
        gameplay_loop()




# Lite logistik

bullet = randrange(1, 5)
blank = 6 - bullet
total = blank + bullet
chamber = ["live"] * bullet + ["blank"] * blank # Listan är nu alla live på på rad först och sedan blank
random.shuffle(chamber) # Här randomiseras listan


## Introsekvensen
delay_print(f"Welcome to RUSSIAN ROULETTE!")

delay_print(f"Before we start... What's your name?")
name = input("> ")
opponent = 33
delay_print(f"...And how old are you, {name}?")
age = input_int("> ")

delay_print(f"So... {name}... Do you know the rules to this game? y/n")
rules = input("> ").lower()

if rules == "y":
  delay_print(f"Good. You need no introduction.")
elif rules == "n":
  delay_print(f"How unfortunate {name}...")

delay_print(f"Let the game begin!")
sleep(4)
clear_output()


# Cheeky onödig loading-screen
msg = "loading."

for i in range(4):
  delay_print(msg)
  msg += "."
  sleep(0.7)

clear_output()

if age < opponent:
  delay_print(f"???:       {name}, since you're the youngest... You get to start!")
  gameplay_loop()
else:
  delay_print(f"???:       {name}, since you're the oldest... You get to start!")
  gameplay_loop()

## Återvänder efter gameplayloop.
delay_print("Would you like to play again? y/n")
retry = input("> \n")

if retry == "y":
  bullet = randrange(1, 5)
  blank = 6 - bullet
  total = blank + bullet
  chamber = ["live"] * bullet + ["blank"] * blank
  random.shuffle(chamber)
  clear_output()
  delay_print(f"???:       {name}? Why did YOU get a second chance?")
  delay_print("???:       oh well...\n")
  sleep(2)
  gameplay_loop()
elif retry == "n":
  exit()
else:
  exit()










```
