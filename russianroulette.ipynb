```python
import random
from random import randrange
import math
from IPython.display import clear_output
from time import sleep


## Self-Explanatory, min gameplay loop. Den går runt tills användaren väljer att sluta spela.

def gameplay_loop():
  global blank
  random.shuffle(chamber)
  print(f"???:       In the gun there are {bullet} bullets and {blank} blanks arranged in an unknown order.")
  print(f"???:       What do you wish to do?\n\n [1] Shoot yourself\n [2] Shoot opponent\n")
  choice = input("> ")

  ## chance = bullet / total





  ## Logistik

  if choice == "1":
    current_round = chamber.pop(0)
    print(f"Very brave")
    print(f"You point the gun at yourself...")
    sleep(1.4)
    if current_round == "live":
      print(f"...BANG")
      print(f"\033[91m --- you died... ---\033[0m")
      sleep(1.4)
      clear_output()
      print(f"Do you want to try again? y/n")
      retry = input("> ")
      if retry == "y":
        gameplay_loop()
      elif retry == "n":
        exit()
    else:
      print("...CLICK! It was a blank")
      blank = blank - 1
      total = blank + bullet
      print(f"You get another chance. There are {total} bullets left in the chamber.")
      gameplay_loop()

  elif choice == "2":
    current_round = chamber.pop(0)
    print(f"Cowardly... or clever?")
    print(f"You point the gun at the opponent...")
    sleep(1.4)
    if current_round == "live":
      print(f"...BANG! You shot the opponent.")
      print(f"CONGRATULATIONS! You win!")
    else:
      print(f"...CLICK! It was a blank. Opponent's turn.")
      sleep(1.4)
      print(f"...BANG!")
      print(f"\033[91m --- you died... ---\033[0m")
      sleep(2)
      clear_output()
      print(f"Do you want to try again? y/n")
      retry = input("> ")
      if retry == "y":
        gameplay_loop()
      elif retry == "n":
        exit()



## Ladda logistiken

bullet = randrange(1, 5)
blank = 6 - bullet
total = blank + bullet
chamber = []
chamber = ["live"] * bullet + ["blank"] * blank
random.shuffle(chamber)


## Lite intro-sekvens

print(f"Welcome to RUSSIAN ROULETTE!")

print(f"Before we start... What's your name?")
name = input("> ")
print(f"Greetings {name}, do you know the rules to this game? y/n")


# Cheeky hacks :)
if name.lower() == "admin":
  print(chamber)

rules = input().lower()

if rules == "y":
  print(f"Good. You need no introduction.")
elif rules == "n":
  print(f"How unfortunate {name}...")
else:
  print("are u dum?")

print(f"Let the game begin!")
sleep(4)
clear_output()

## Useless loading screen.

msg = "loading."

for i in range(4):
  print(msg)
  msg += "."
  sleep(0.7)

clear_output()
print(f"???:       {name} since you're the youngest... You get to start!")
gameplay_loop()

print("Would you like to play again? y/n")
retry = input("> ")

if retry == "y":
  random.shuffle(chamber)
  gameplay_loop()
elif retry == "n":
  exit()
else:
  print("haha u very dum u no able to spell")
  print("Deleting System32...")
  sleep(2)
  exit()








```
