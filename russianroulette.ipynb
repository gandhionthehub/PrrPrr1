```python
import random
from random import randrange
import math
from IPython.display import clear_output
from time import sleep



# Snyggare delay_print kommando
def delay_print(text, delay=0.1):
    for letter in text:
        print(letter, end="", flush=True)
        sleep(0.05)
    print()

# Spelsekvensen
def gameplay_loop():
  global blank
  delay_print(f"???:       In the gun there are {bullet} bullets and {blank} blanks arranged in an unknown order.")
  delay_print(f"???:       What do you wish to do?\n\n [1] Shoot yourself\n [2] Shoot opponent\n")
  choice = input("> ")

  ## chance = bullet / total





  ## Logistik

  if choice == "1":
    current_round = chamber.pop(0)
    delay_print(f"Very brave")
    delay_print(f"You point the gun at yourself...")
    sleep(1.4)
    if current_round == "live":
      delay_print(f"...BANG")
      delay_print(f"\033[91m --- you died... ---\033[0m")
      sleep(1.4)
      clear_output()
      delay_print(f"Do you want to try again? y/n")
      retry = input("> ")
      if retry == "y":
        gameplay_loop()
      elif retry == "n":
        exit()
    else:
      delay_print("...CLICK! It was a blank")
      blank = blank - 1
      total = blank + bullet
      delay_print(f"You get another chance. There are {total} bullets left in the chamber.")
      gameplay_loop()

  elif choice == "2":
    current_round = chamber.pop(0)
    delay_print(f"Cowardly... or clever?")
    delay_print(f"You point the gun at the opponent...")
    sleep(1.4)
    if current_round == "live":
      delay_print(f"...BANG! You shot the opponent.")
      delay_print(f"CONGRATULATIONS! You win!")
    else:
      delay_print(f"...CLICK! It was a blank. Opponent's turn.")
      sleep(1.4)
      delay_print(f"...BANG!")
      delay_print(f"\033[91m --- you died... ---\033[0m")
      sleep(1.4)
      clear_output()
      delay_print(f"Would you like to try again? y/n")
      retry = input("> ")
      if retry == "y":
        clear()
        gameplay_loop()
      elif retry == "n":
        exit()



# Lite logistik

bullet = randrange(1, 5)
blank = 6 - bullet
total = blank + bullet
chamber = ["live"] * bullet + ["blank"] * blank
random.shuffle(chamber)


## Introsekvensen
delay_print(f"Welcome to RUSSIAN ROULETTE!")

delay_print(f"Before we start... What's your name?")
name = input("> ")
delay_print(f"Greetings {name}, do you know the rules to this game? y/n")
rules = input().lower()

if rules == "y":
  delay_print(f"Good. You need no introduction.")
elif rules == "n":
  delay_print(f"How unfortunate {name}...")

delay_print(f"Let the game begin!")
sleep(4)
clear_output()


# Cheeky onödig loading-screen
msg = "loading."

for i in range(4):
  delay_print(msg)
  msg += "."
  sleep(0.7)

clear_output()
delay_print(f"???:       {name} since you're the youngest... You get to start!")
gameplay_loop()

## Återvänder efter gameplayloop.

delay_print("Would you like to play again? y/n")
retry = input("> ")

if retry == "y":
  random.shuffle(chamber)
  gameplay_loop()
elif retry == "n":
  exit()
else:
  exit()









```
